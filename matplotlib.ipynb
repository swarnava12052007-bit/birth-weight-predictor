```python
plt.plot(x, np.sin(x))
```

```python
plt.plot(x, np.sin(x), linestyle = "--", color = "r", linewidth = 4, label = "sin(x)")
plt.plot(x, np.cos(x), linestyle = "--", color = "b", linewidth = 4, label = "cos(x)")
plt.title("Trigonometric functions")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid()
plt.show()
```

```python
#for two plots in one page
plt.figure() #initialising a page to make plot
plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x))

plt.subplot(1, 2, 2)
plt.plot(x, np.cos(x))

plt.show()
```

```python
plt.figure() #initialising a page to make plot
plt.subplot(2, 1, 1)
plt.plot(x, np.sin(x))

plt.subplot(2, 1, 2)
plt.plot(x, np.cos(x))

plt.show()
```

```python
plt.figure()
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(x, np.tan(x))
plt.grid()

plt.subplot(2, 2, 4)
plt.plot(x, np.sin(x+1))
plt.grid()

plt.show()
```

```python
#histogram>> continuous numerical data

data = np.random.randn(1000) #by default 10 bins are created
plt.hist(data)
plt.show()
```

```python
data1 = np.random.randn(1000)
plt.hist(data1, color = 'r', bins = 20)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Histogram of Random Data")
plt.grid()
plt.show()
```

```python
#stacked histogram >> two histograms together
data10 = np.random.normal(0, 1, 5000)
data20 = np.random.normal(3, 2, 5000)
```

```python
plt.hist([data10, data20], color = ['r', 'b'], bins = 30, label = ['data10', 'data20'])
plt.title("Stacked Histogram")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid()
plt.show()
```

```python
#bar plot >> To see the districution of categorical variables
x = ["a", "b", "c","d"] #name of students
y = [4, 6, 1, 10] #no of days present in the class

plt.bar(x, y, color = 'r') #vertical plot
```

```python
x = ["a", "b", "c","d"] #name of students
y = [4, 6, 1, 10] #no of days present in the class

plt.barh(x, y, color = 'r') #horizontal plot
```

```python
#pie char >> To understand the percent distribution of data
langs = ["C", "Python", "Java"]
students = [20, 100, 40]

plt.pie(students, labels = langs, autopct = "%1.1f%%")
plt.title("Pie chart")
plt.show()
```

```python
#focus on one segment
explode = (0.0, 0.1, 0.1)
langs = ["C", "Python", "Java"]
students = [20, 100, 40]

plt.pie(students, labels = langs, autopct = "%1.1f%%", explode = explode)
plt.title("Pie chart")
plt.show()
```

```python
explode = (0.0, 0.1, 0.1)
langs = ["C", "Python", "Java"]
students = [20, 100, 40]

plt.pie(students, labels = langs, autopct = "%1.1f%%", explode = explode, shadow = True)
plt.title("Pie chart")
plt.show()
```

```python
#3d plot >> 3 dimesion plot
x = np.random.rand(20)
y = np.random.rand(20)
z = np.random.rand(20)

fig = plt.figure()
ax = fig.add_subplot(projection = '3d')
ax.scatter(x, y, z, c='red')
plt.show()
```

```python
#area plot >> to see the cumulative distribution of data
years = [2021, 2018, 2022, 2023, 2017]
category1 = [20, 25, 30, 28, 35]
category2 = [10, 15, 20, 18, 25]
category3 = [5, 8, 10, 9, 15]

plt.stackplot(years, category1, category2, category3, labels = ["category1", "category2", "category3",],
             colors = ["skyblue","lightgreen", "orange"])
plt.xlabel("years")
plt.ylabel("Values")
plt.title("Area plot")
plt.legend()
plt.show()
```
