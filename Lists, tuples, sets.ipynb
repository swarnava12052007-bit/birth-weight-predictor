```python
#task 1 
gum_brands = {"Orbit", "Extra", "5 Gum", "Juicy Fruit", "Doublemint", "Orbit"}

print(gum_brands)
print(type(gum_brands))
```

```python
#task 2
gum_brands = {"Orbit", "Extra", "5 Gum", "Juicy Fruit", "Doublemint"}
gum_brands.add("Big Red")
print(gum_brands)
```

```python
#task 3
sugar_free_gums = {"Orbit", "Extra", "5 Gum", "Eclipse"}
print(gum_brands.difference(sugar_free_gums)) 
```

```python
#task 4
print(gum_brands.intersection(sugar_free_gums))
```

```python
#task 5 
print(gum_brands.union(sugar_free_gums))
```

```python
#task 6 
print(gum_brands.symmetric_difference(sugar_free_gums))
```

```python
#task 7
gum_brands.discard("Juicy Fruit")
print(gum_brands)
```

```python
#task 8
classic_gums = {"Doublemint", "Juicy Fruit"}
print(classic_gums.issubset(gum_brands))
print(gum_brands.issuperset(classic_gums))
```
