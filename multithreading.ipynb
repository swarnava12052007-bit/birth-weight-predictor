```python
import time
start = time.perf_counter()

def classes():
    print("doing something")
    print("still doing")
    time.sleep(1)
    print("start again")
classes()

end = time.perf_counter()

print(f"this task was finished in {round(end-start, 2)} seconds")
```

```python
import time
import threading

start = time.perf_counter()

def class1():
    print("start again")
    print("keep going")
    print("cannot stop")
    time.sleep(1)
    print("work again")

t1 = threading.Thread(target = class1)
t2 = threading.Thread(target = class1)

t1.start()
t2.start()

t1.join()
t2.join()

end = time.perf_counter()

print(f"to be ended in {round(end - start, 2)} seconds")
```

```python
import time
import threading #python module
start = time.perf_counter()

def test_func():
    print("do something")
    print("sleep for 1 sec")
    time.sleep(1)
    print("done with sleeping")

threads = []
for i in range(10):
    t = threading.Thread(target = test_func)
    t.start()
    threads.append(t)

for thread in threads:
    thread.join()


end = time.perf_counter()

print(f"The program finished in {round(end-start, 2)} seconds")
```

```python
#shared variable across all the threads

start = time.perf_counter()
shared_counter = 0
counter_lock = threading.Lock() #locking the counter for a specific thread

def increment_shared_counter(x):
    global shared_counter #that can be accessed by all the threads
    with counter_lock:
        shared_counter = shared_counter+1
        print(f"Thread {x}: incremented shared counter to {shared_counter}")
        time.sleep(1)

threads = [threading.Thread(target = increment_shared_counter, args = (i,)) for i in [1, 2, 3, 4, 5, 6]]

for thread in threads:
    thread.start()
for thread in threads:
    thread.join()

end = time.perf_counter()
print(f"The program finished in {round(end-start, 2)} seconds.")
```
