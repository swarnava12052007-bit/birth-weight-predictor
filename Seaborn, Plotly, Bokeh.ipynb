```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
flowers = sns.load_dataset('iris')
```

```python
from bokeh.sampledata.iris import flowers
from bokeh.plotting import figure, output_file, show
```

```python
flowers
```

```python
output_notebook()
p = figure(title = 'test_flower')
p.xaxis.axis_label = "petal_length"
p.yaxis.axis_label = "petal_width"
p.scatter(flowers['petal_length'], flowers['petal_width'], size=10) #circle for scatter plot
show(p)
```

```python
output_file("test.html")
p = figure(title = 'test_flower')
p.xaxis.axis_label = "petal_length"
p.yaxis.axis_label = "petal_width"
p.line(flowers['petal_length'], flowers['petal_width']) #line for line chart
show(p)
```
