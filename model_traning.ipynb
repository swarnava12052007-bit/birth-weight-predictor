```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split as tts
```

```python
df
```

```python
#spliting the data into x and y
y = df["bwt"]
y
```

```python
x=df.iloc[:,1:]
x.head()
```

```python
#train test split
from sklearn.model_selection import train_test_split as tts
```

```python
x_train, x_test, y_train, y_test = tts(x,y, test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge
```

```python
lr = LinearRegression()
lasso = Lasso()
ridge = Ridge()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
x_train = pd.DataFrame(imputer.fit_transform(x_train), columns=x_train.columns)
x_test = pd.DataFrame(imputer.transform(x_test), columns=x_test.columns)
```

```python
lr.fit(x_train, y_train)
lasso.fit(x_train, y_train)
ridge.fit(x_train, y_train)
```

```python
lr_pred = lr.predict(x_test)
lasso_pred = lasso.predict(x_test)
ridge_pred = ridge.predict(x_test)
```

```python
#model evaluation
from sklearn.metrics import mean_squared_error, r2_score
```

```python
lr_rscore = r2_score(y_test, lr_pred)
lasso_rscore = r2_score(y_test, lasso_pred)
ridge_rscore = r2_score(y_test, ridge_pred)
```

```python
lr_mse = mean_squared_error(y_test, lr_pred)
lasso_mse = mean_squared_error(y_test, lasso_pred)
ridge_mse = mean_squared_error(y_test, ridge_pred)
```

```python
print("\nPerformance of Linear Regression Model:")
print("r2_score:", lr_rscore)
print("mse:", lr_mse)

print("\nPerformance of Lasso Regression Model:")
print("r2_score:", lasso_rscore)
print("mse:", lasso_mse)
      
print("\nPerformance of Ridge Regression Model:")
print("r2_score:", ridge_rscore)
print("mse:", ridge_mse)
```
